```python
# Label encoding
from sklearn.preprocessing import LabelEncoder
lab=LabelEncoder()
```

```python
df['education']=lab.fit_transform(df['education'])
df['self_employed']=lab.fit_transform(df['self_employed'])
df
```

```python
#separate input and output
x=df.iloc[:,:-1].values
x
```

```python
x.ndim
```

```python
y=df.iloc[:,-1].values
y
```

```python
y.ndim
```

```python

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=30,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
Scalar=StandardScaler()
Scalar.fit(x_train)
x_train=Scalar.transform(x_train)
x_test=Scalar.transform(x_test)
x_train
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(criterion='entropy')
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
lst=[dt,rf]
for i in lst: 
    print("Model name is",i)
    i.fit(x_train,y_train)
    y_pred=i.predict(x_test)
    y_pred
    print("***********************")
    print("Model Confusion metrics is")
    print(confusion_matrix(y_test,y_pred)) 
    print ("****************") 
    print("Accuracy score is")
    print(accuracy_score(y_test,y_pred))
    print("******************")
    print(classification_report(y_test,y_pred))
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    criterion='entropy',
    max_depth=5,
    min_samples_split=20,
    min_samples_leaf=10,
    random_state=42
)
dt = DecisionTreeClassifier(
    criterion="entropy",      # or "gini"
    max_depth=5,              # prevents overfitting
    min_samples_split=20,
    min_samples_leaf=10,
    random_state=42
)
```

```python
lst=[dt,rf]
for i in lst: 
    print("Model name is",i)
    i.fit(x_train,y_train)
    y_pred1=i.predict(x_test)
    y_pred1
    print("***********************")
    print("Model Confusion metrics is")
    print(confusion_matrix(y_test,y_pred1)) 
    print ("****************") 
    print("Accuracy score is")
    print(accuracy_score(y_test,y_pred1))
    print("******************")
    print(classification_report(y_test,y_pred1))
```

```python
model= RandomForestClassifier(
    n_estimators=100,
    criterion='entropy',
    max_depth=5,
    min_samples_split=20,
    min_samples_leaf=10,
    random_state=42
)
```

```python
model.fit(x_train,y_train)
y_pred2=model.predict(x_test)
y_pred2
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,ConfusionMatrixDisplay
cm=confusion_matrix(y_test,y_pred2)
cm
```

```python
lab=[1,0]
cmd=ConfusionMatrixDisplay(cm,display_labels=lab)
cmd.plot()
```

```python
score=accuracy_score(y_pred2,y_test)
score
```
